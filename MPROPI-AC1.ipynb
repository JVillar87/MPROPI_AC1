## Repte final: Validador de contrasenyes (15min)

Implementa una funció anomenada `validar_contrasenya()` que rebi una cadena de text i retorni `True` si la contrasenya és vàlida o `False` en cas contrari.

Una contrasenya és vàlida si compleix **totes** aquestes condicions:

1. Té com a mínim **8 caràcters**.
2. Conté almenys **un número**.
3. Conté almenys **una lletra majúscula**.
4. No conté cap espai.

### Exemples

```python
validar_contrasenya("Hola1234")   # True
validar_contrasenya("hola1234")   # False
validar_contrasenya("HOLAabcd")   # False
validar_contrasenya("Ho 123456")  # False
validar_contrasenya("Hola12")     # False

````python
# Repte final
def validar_contrasenya(contrasenya):
    te_numero = False
    te_majuscula = False
    te_espai = False

    # Escriu aquí la teva solució
    for caracter in contrasenya:
        if caracter.isdigit():
            te_numero = True
        if caracter.isupper():
            te_majuscula = True
        if caracter == " ":
            te_espai = True

        if len(contrasenya) < 8:
            return False
        if not te_numero:
            return False
        if not te_majuscula:
            return False
        if te_espai:
            return False
    return True
    pass

#Proves

print(validar_contrasenya("Hola1234"))
print(validar_contrasenya("hola1234"))
print(validar_contrasenya("HOLAabcd"))
print(validar_contrasenya("Ho 123456"))
print(validar_contrasenya("Hola12"))
````
