### Exercici 0.1
Amb `a = 10` i `b = 3`, calcula suma, resta, producte, divisió real, divisió entera, residu i potència. Mostra també el tipus de cada resultat.

````sql
# Escriu aquí la teva solució
a = 10
b = 3
#suma
c = a + b
print(c, type(c))
#resta
d = a - b
print(d, type(d))
#multiplicació
e = a * b
print(e, type(e))
#divisió real
f = a / b
print(f, type(f))
#divisió entera
h = a // b
print(h, type(h))
#residu
i = a % b
print(i, type(i))
#potència
j = a ** b
print(j, type(j))
#producte
k = a * b
print(k, type(k))
````
